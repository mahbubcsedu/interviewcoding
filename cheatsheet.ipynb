## Inplace array missing numbers finding
LC-448
LC-42

```python
from typing import List
def findAllMissingNumber(nums:List[int])->List[int]:
    for num in nums:
            index = abs(num)-1
            nums[index] = -abs(nums[index])

        res = []

        for i in range(len(nums)):
            if nums[i] >=0:
                res.append(i+1)
        return res
```

## 2. Core Python Implementations & Templates

```python
# Binary Search (Template for finding exact match or boundary)
def binary_search(arr: list[int], target: int) -> int:
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
```

```python
# Cyclic Sort Pattern (Useful for LC-448, LC-41, finding missing numbers in range [1, n])
def cyclic_sort(nums: list[int]) -> list[int]:
    i = 0
    while i < len(nums):
        correct_idx = nums[i] - 1
        # Swapping element to its correct position if it lies within [1, n]
        if 0 <= correct_idx < len(nums) and nums[i] != nums[correct_idx]:
            nums[i], nums[correct_idx] = nums[correct_idx], nums[i]
        else:
            i += 1
    return nums
```

```python
# Monotonic Queue (Sliding Window Maximum template)
from collections import deque

def max_sliding_window(nums: list[int], k: int) -> list[int]:
    dq = deque()  # stores indices of elements in decreasing order
    res = []
    for i, num in enumerate(nums):
        # Keep elements in queue in decreasing order
        while dq and nums[dq[-1]] < num:
            dq.pop()
        dq.append(i)
        # Remove indices out of current window
        if dq[0] == i - k:
            dq.popleft()
        # Add maximum for current window to result
        if i >= k - 1:
            res.append(nums[dq[0]])
    return res
```

```python
# Breadth-First Search (BFS) for Graph Connectivity/Shortest Path
def bfs(graph: dict, start: int) -> list[int]:
    visited = set([start])
    queue = deque([start])
    order = []
    while queue:
        node = queue.popleft()
        order.append(node)
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return order
```

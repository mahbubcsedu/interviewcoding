## 1. Linear DP

### Theory & Core Concept
Linear DP is the most fundamental form of DP where the state space can be organized in a single dimension (typically an array index $i$). The transition to state $i$ depends on a constant number of previous states (e.g., $i-1$, $i-2$).

* **State Representation:** `dp[i]` = optimal solution for the prefix of size $i$.
* **Transition:** `dp[i] = transition(dp[i-1], dp[i-2], ...)`

### References & Resources
* [GeeksforGeeks: Introduction to Linear DP](https://www.geeksforgeeks.org/introduction-to-dynamic-programming/)

### Sample Problem Solution: 70. Climbing Stairs
```python
def climbStairs(n: int) -> int:
    if n <= 2:
        return n
    prev2, prev1 = 1, 2
    for _ in range(3, n + 1):
        curr = prev1 + prev2
        prev2 = prev1
        prev1 = curr
    return prev1
```

## 2. Knapsack

### Theory & Core Concept
Knapsack DP deals with selecting items with weights and values to maximize total value without exceeding a capacity limit.
* **0/1 Knapsack:** Items cannot be repeated: `dp[i][w] = max(dp[i-1][w], dp[i-1][w-weight[i]] + value[i])`.
* **Unbounded Knapsack:** Items can be reused infinitely.

### References & Resources
* [CP-Algorithms: Knapsack](https://cp-algorithms.com/dynamic_programming/knapsack.html)

### Sample Problem Solution: 322. Coin Change
```python
def coinChange(coins: list[int], amount: int) -> int:
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    for coin in coins:
        for x in range(coin, amount + 1):
            dp[x] = min(dp[x], dp[x - coin] + 1)
    return dp[amount] if dp[amount] != float('inf') else -1
```

## 3. Multi-Dimensional DP

### Theory & Core Concept
Multi-Dimensional DP uses states parameterized by two or more variables (e.g., `dp[i][j][k]`), common when matching strings, moving on grids, or tracking state machine constraints.

### Sample Problem Solution: 120. Triangle
```python
def minimumTotal(triangle: list[list[int]]) -> int:
    dp = triangle[-1]
    for row in range(len(triangle) - 2, -1, -1):
        for col in range(len(triangle[row])):
            dp[col] = triangle[row][col] + min(dp[col], dp[col + 1])
    return dp[0]
```

## 4. Interval DP

### Theory & Core Concept
Interval DP solves problems on subarrays or intervals $[i, j]$ by partitioning into subsegments $[i, k]$ and $[k+1, j]$.
* **Transition:** `dp[i][j] = min/max(dp[i][k] + dp[k+1][j] + cost)` for $i \le k < j$.

### Sample Problem Solution: 312. Burst Balloons
```python
def maxCoins(nums: list[int]) -> int:
    A = [1] + [x for x in nums if x > 0] + [1]
    n = len(A)
    dp = [[0] * n for _ in range(n)]
    for l in range(1, n - 1):
        for i in range(1, n - l):
            j = i + l - 1
            for k in range(i, j + 1):
                dp[i][j] = max(dp[i][j], dp[i][k-1] + A[i-1]*A[k]*A[j+1] + dp[k+1][j])
    return dp[1][n-2]
```

## 7. DP on Trees

### Theory & Core Concept
Optimization performed bottom-up on tree topologies, typically using post-order depth-first search (DFS).

### Sample Problem Solution: 337. House Robber III
```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def rob(root: TreeNode) -> int:
    def dfs(node):
        if not node: return (0, 0)
        left, right = dfs(node.left), dfs(node.right)
        rob_curr = node.val + left[1] + right[1]
        no_rob_curr = max(left) + max(right)
        return (rob_curr, no_rob_curr)
    return max(dfs(root))
```

## 8. String DP

### Theory & Core Concept
Involves finding patterns, distances, or edits between prefixes of strings.

### Sample Problem Solution: 1143. Longest Common Subsequence
```python
def longestCommonSubsequence(text1: str, text2: str) -> int:
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    return dp[m][n]
```

## 10. Classic DPs

### A. Kadane’s Algorithm
```python
def maxSubArray(nums: list[int]) -> int:
    curr_max = global_max = nums[0]
    for x in nums[1:]:
        curr_max = max(x, curr_max + x)
        global_max = max(global_max, curr_max)
    return global_max
```

### B. Longest Increasing Subsequence (LIS)
```python
import bisect
def lengthOfLIS(nums: list[int]) -> int:
    sub = []
    for x in nums:
        idx = bisect.bisect_left(sub, x)
        if idx == len(sub): sub.append(x)
        else: sub[idx] = x
    return len(sub)
```

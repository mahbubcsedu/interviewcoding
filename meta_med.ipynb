## 167. Two Sum II - Input Array Is Sorted

Given a 1-indexed array of integers numbers that is already sorted in non-decreasing order, find two numbers such that they add up to a specific target number. Let these two numbers be numbers[index1] and numbers[index2] where 1 <= index1 < index2 < numbers.length.

Return the indices of the two numbers, index1 and index2, added by one as an integer array [index1, index2] of length 2.

The tests are generated such that there is exactly one solution. You may not use the same element twice.

Your solution must use only constant extra space.

 ```python

Example 1:

Input: numbers = [2,7,11,15], target = 9
Output: [1,2]
Explanation: The sum of 2 and 7 is 9. Therefore, index1 = 1, index2 = 2. We return [1, 2].
Example 2:

Input: numbers = [2,3,4], target = 6
Output: [1,3]
Explanation: The sum of 2 and 4 is 6. Therefore index1 = 1, index2 = 3. We return [1, 3].
Example 3:

Input: numbers = [-1,0], target = -1
Output: [1,2]
Explanation: The sum of -1 and 0 is -1. Therefore index1 = 1, index2 = 2. We return [1, 2].
```

### 💡 Interview Framework: Two Sum II (Sorted Array)

#### 1. Edge Cases to Consider
- **Target Not Found**: While the problem description typically guarantees one solution, it's good practice to discuss what to return if no pair adds up to the target (e.g., `[-1, -1]`).
- **Negative Numbers**: Elements like `[-5, -3, 0, 2, 4]` with a target of `-1` (e.g. `[-3, 2]`). The logic remains identical.
- **Smallest Possible Input**: An array of size exactly `2` (the absolute minimum since we need a pair).

#### 2. Alternative Approaches & Trade-offs
- **Brute Force (Nested Loops)**:
  - *Time Complexity*: $O(N^2)$
  - *Space Complexity*: $O(1)$
- **Hash Map**:
  - *Time Complexity*: $O(N)$
  - *Space Complexity*: $O(N)$ (Fails the constant space $O(1)$ constraint).
- **Binary Search** (for each element `nums[i]`, search for `target - nums[i]` in the remainder of the array):
  - *Time Complexity*: $O(N \log N)$
  - *Space Complexity*: $O(1)$
- **Two-Pointer Approach (Optimal)**:
  - *Time Complexity*: $O(N)$ single pass.
  - *Space Complexity*: $O(1)$ constant space.

````python
def twoSumII(nums, target):
  # Two-pointer approach: O(n) time complexity, O(1) space complexity.
  # Note: The problem requires returning 1-indexed results.
  left, right = 0, len(nums) - 1

  while left < right:
    two_sum = nums[left] + nums[right]

    if two_sum == target:
      return [left + 1, right + 1] # Converted to 1-indexed
    elif two_sum < target:
      left += 1
    else:
      right -= 1
  return [-1, -1]
````

## 198 House Robber
You are a professional robber planning to rob houses along a street. Each house has a certain amount of money stashed, the only constraint stopping you from robbing each of them is that adjacent houses have security systems connected and it will automatically contact the police if two adjacent houses were broken into on the same night.

Given an integer array nums representing the amount of money of each house, return the maximum amount of money you can rob tonight without alerting the police.

 ```python

Example 1:

Input: nums = [1,2,3,1]
Output: 4
Explanation: Rob house 1 (money = 1) and then rob house 3 (money = 3).
Total amount you can rob = 1 + 3 = 4.
Example 2:

Input: nums = [2,7,9,3,1]
Output: 12
Explanation: Rob house 1 (money = 2), rob house 3 (money = 9) and rob house 5 (money = 1).
Total amount you can rob = 2 + 9 + 1 = 12.
```

### 💡 Interview Framework: House Robber

#### 1. Edge Cases to Consider
- **Zero Houses (`len(nums) == 0`)**: The maximum possible loot is `0`.
- **One House (`len(nums) == 1`)**: Return the value of that single house.
- **Two Houses (`len(nums) == 2`)**: The optimal choice is simply `max(nums[0], nums[1])`.
- **Empty / Zero loot values**: e.g., `[0, 0, 0]`. Returns `0`.

#### 2. Alternative Approaches & Trade-offs
- **Recursion without Memoization**:
  - *Time Complexity*: $O(2^N)$
  - *Space Complexity*: $O(N)$ due to call stack.
- **Top-Down Dynamic Programming (Recursion + Memoization)**:
  - *Time Complexity*: $O(N)$
  - *Space Complexity*: $O(N)$ for both memoization array and call stack.
- **Bottom-Up Space-Optimized DP (Optimal)**:
  - *Time Complexity*: $O(N)$ single pass.
  - *Space Complexity*: $O(1)$ by keeping track of only the previous two states (`rob1` and `rob2`).

````python
def rob(nums: list[int]) -> int:
  # Iterative DP with O(1) space complexity.
  # rob = max(nums[i] + rob_i_minus_2, rob_i_minus_1)
  rob1, rob2 = 0, 0

  for num in nums:
    new_rob = max(rob1 + num, rob2)
    rob1 = rob2
    rob2 = new_rob

  return rob2

# Example usage
print(rob([2, 7, 9, 3, 1])) # Output: 12
````

## 1209. Remove All Adjacent Duplicates in String II

You are given a string s and an integer k, a k duplicate removal consists of choosing k adjacent and equal letters from s and removing them, causing the left and the right side of the deleted substring to concatenate together.

We repeatedly make k duplicate removals on s until we no longer can.

Return the final string after all such duplicate removals have been made. It is guaranteed that the answer is unique.

 ```

Example 1:

Input: s = "abcd", k = 2
Output: "abcd"
Explanation: There's nothing to delete.
Example 2:

Input: s = "deeedbbcccbdaa", k = 3
Output: "aa"
Explanation:
First delete "eee" and "ccc", get "ddbbbdaa"
Then delete "bbb", get "dddaa"
Finally delete "ddd", get "aa"
Example 3:

Input: s = "pbbcggttciiippooaais", k = 2
Output: "ps"
```

### 💡 Interview Framework: Remove All Adjacent Duplicates II

#### 1. Edge Cases to Consider
- **String Length less than k**: Cannot have any deletions; return original string.
- **Repeated deletions cascading**: e.g., `s = "abbba", k = 3`. Deleting `bbb` yields `aa`. No further deletion can happen.
- **k = 1**: Deletes every single character immediately, resulting in an empty string.

#### 2. Alternative Approaches & Trade-offs
- **Brute Force (String Reconstruction)**: Repeatedly scan the string to find $k$ duplicate characters, slice them out, and repeat.
  - *Time Complexity*: $O(N^2 / k)$
  - *Space Complexity*: $O(N)$ or $O(N^2)$ depending on copy overhead.
- **Stack of Characters & Counts (Optimal)**:
  - *Time Complexity*: $O(N)$
  - *Space Complexity*: $O(N)$

````python
'''
aacbbccd k=2

i=0, stack[[a,1]]
i=1, stack[-1][0] == s[i], so, instead of insert, delete top of the entry, so stack=[]
i=2, empty stack, so enter [c, 1]
i=3, doest not match top, stack=[[c,1],[b ,1]]
i=4, match top of the stack, so, run k-1 times pop, stack[[c,1]]
.....
.....
'''

def removeDuplicates(s: str, k: int) -> str:
    stack=[]

    for i in range(len(s)):
        if stack and stack[-1][0]==s[i] and stack[-1][1]==k-1:
            for i in range(k-1):
                stack.pop()
        elif stack and stack[-1][0]==s[i]:
            stack.append((s[i], stack[-1][1]+1))
        else:
            stack.append((s[i],1))
    return "".join(c for c,i in stack)
print(removeDuplicates("aacbbccd", 2))
````

````python
def removeDuplicate(s: str, k: int) -> str:
  # Using a stack to track [char, frequency count]
  # Time: O(n), Space: O(n)
  stack = []

  for ch in s:
    if stack and stack[-1][0] == ch:
      stack[-1][1] += 1
      if stack[-1][1] == k:
        stack.pop()
    else:
      stack.append([ch, 1])

  return "".join(ch * count for ch, count in stack)

print(removeDuplicate("deeedbbcccbdaa", 3)) # Output: "aa"
````

## 102
Binary Tree Level Order Traversal


### 💡 Interview Framework: Binary Tree Level Order Traversal

#### 1. Edge Cases to Consider
- **Empty Tree (`root = None`)**: Return an empty list `[]` immediately.
- **Single Node**: Return `[[root.val]]`.
- **Skewed Tree (Left or Right)**: Essentially acts like a linked list, ensure depth processing remains robust.

#### 2. Alternative Approaches & Trade-offs
- **Breadth-First Search (Queue - Optimal)**:
  - *Time Complexity*: $O(N)$ since each node is visited exactly once.
  - *Space Complexity*: $O(D)$ where $D$ is the maximum diameter of the tree (leaf level can hold up to $N/2$ nodes).
- **Depth-First Search (DFS Recursion)** (passing level index):
  - *Time Complexity*: $O(N)$
  - *Space Complexity*: $O(H)$ where $H$ is tree height (recursion stack).

````python
import collections
class Node:
  def __init__(self, val, left=None, right=None):
    self.val, self.left, self.right = val, left, right


def btreeLevelOrder(root):
  q=collections.deque()
  res=[]
  q.append(root)
  # res.append([root.val])

  while q:
    level = []

    for _ in range(len(q)):
      node = q.popleft()
      level.append(node.val)

      if node.left:
        q.append(node.left)
      if node.right:
        q.append(node.right)

    res.append(level)
    level = []
  return res


tree = Node(1, Node(2, Node(3, None, Node(4, None))), Node(7, Node(0), Node(8, Node(9, Node(10)))))
print(btreeLevelOrder(tree))
````

## 1123. Lowest Common Ancestor of Deepest Leaves

Given the root of a binary tree, return the lowest common ancestor of its deepest leaves.

Recall that:

The node of a binary tree is a leaf if and only if it has no children
The depth of the root of the tree is 0. if the depth of a node is d, the depth of each of its children is d + 1.
The lowest common ancestor of a set S of nodes, is the node A with the largest depth such that every node in S is in the subtree with root A.

```
Example 1:


Input: root = [3,5,1,6,2,0,8,null,null,7,4]
Output: [2,7,4]
Explanation: We return the node with value 2, colored in yellow in the diagram.
The nodes coloured in blue are the deepest leaf-nodes of the tree.
Note that nodes 6, 0, and 8 are also leaf nodes, but the depth of them is 2, but the depth of nodes 7 and 4 is 3.
Example 2:

Input: root = [1]
Output: [1]
Explanation: The root is the deepest node in the tree, and it's the lca of itself.
Example 3:

Input: root = [0,1,3,null,2]
Output: [2]
Explanation: The deepest leaf node in the tree is 2, the lca of one node is itself.
```

### 💡 Interview Framework: LCA of Deepest Leaves

#### 1. Edge Cases to Consider
- **Single Node Tree**: The deepest leaf node is the root itself; return the root.
- **Perfectly Balanced Tree**: Deepest leaves are on both extremes. The LCA will be the absolute root.
- **Deepest leaf is unique**: If only one leaf resides at maximum depth, the LCA is that leaf node itself.

#### 2. Alternative Approaches & Trade-offs
- **Two-Pass Approach**: Find maximum depth in pass 1, then find the LCA of all nodes matching this depth in pass 2.
  - *Time Complexity*: $O(N)$
  - *Space Complexity*: $O(H)$
- **One-Pass DFS (Post-order - Optimal)**: Let `dfs(node)` return `(depth, lca)`. We can recursively compute subtree depths and bubble up the LCA node when paths match.
  - *Time Complexity*: $O(N)$
  - *Space Complexity*: $O(H)$ recursion stack space.

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def lca_deepestnodes(root):
  # Depth-first-search post-order traversal.
  # Compares max depth of left & right subtrees dynamically.
  def dfs(node):
    if not node:
      return 0, None

    ld, ln = dfs(node.left)
    rd, rn = dfs(node.right)

    if ld == rd:
      return ld + 1, node
    elif ld > rd:
      return ld + 1, ln
    else:
      return rd + 1, rn

  d, node = dfs(root)
  return node

# Setup using matching class definition
tree = TreeNode(3, TreeNode(5, TreeNode(6), TreeNode(2, TreeNode(7), TreeNode(4))), TreeNode(1, TreeNode(0), TreeNode(8)))
lca = lca_deepestnodes(tree)
print(lca.val) # Output: 2
````

 ```python
Example 3:

Input: nums = [1,2,4], k = 3
Output: 6
Explanation: The missing numbers are [3,5,6,7,...], hence the third missing number is 6.
```

### 💡 Interview Framework: Missing Element in Sorted Array

#### 1. Edge Cases to Consider
- **k is greater than all missing numbers inside the array bounds**: e.g., `nums = [1, 2, 3], k = 5`. The search scope must fall outside the boundaries of the array.
- **No elements missing within the bounds**: e.g., consecutive array `[1, 2, 3, 4]`.
- **Array starts with a high number**: e.g., `[100, 105]`. The missing counting starts relative to the leftmost index `nums[0]`.

#### 2. Alternative Approaches & Trade-offs
- **Linear Scan**: Traverse through the sorted array and count the missing numbers incrementally.
  - *Time Complexity*: $O(N)$
  - *Space Complexity*: $O(1)$
- **Binary Search (Optimal)**: Calculate the number of missing elements at index `mid` as `nums[mid] - nums[0] - mid` to jump sections.
  - *Time Complexity*: $O(\log N)$
  - *Space Complexity*: $O(1)$

````python
def kthMissingNumber(nums, k):
  # Binary Search approach to find the missing element in O(log N) time.
  # Number of missing values before index i is: nums[i] - nums[0] - i
  left, right = 0, len(nums) - 1
  start = nums[0]

  while left <= right:
    mid = (left + right) // 2
    missing = nums[mid] - start - mid

    if missing < k:
      left = mid + 1
    else:
      right = mid - 1

  return start + k + right

print(kthMissingNumber([4, 7, 9, 10], 3)) # Output: 8
````

## 346. Moving Average from Data Stream

Given a stream of integers and a window size, calculate the moving average of all integers in the sliding window.

Implement the MovingAverage class:

MovingAverage(int size) Initializes the object with the size of the window size.
double next(int val) Returns the moving average of the last size values of the stream.

```
Example 1:

Input
["MovingAverage", "next", "next", "next", "next"]
[[3], [1], [10], [3], [5]]
Output
[null, 1.0, 5.5, 4.66667, 6.0]

Explanation
MovingAverage movingAverage = new MovingAverage(3);
movingAverage.next(1); // return 1.0 = 1 / 1
movingAverage.next(10); // return 5.5 = (1 + 10) / 2
movingAverage.next(3); // return 4.66667 = (1 + 10 + 3) / 3
movingAverage.next(5); // return 6.0 = (10 + 3 + 5) / 3
```

````python
class MovingAverage:

    def __init__(self, size: int):
        self.size=size
        self.window = collections.deque()
        self.n_of_items=0
        self.total=0.0


    def next(self, val: int) -> float:
        if self.n_of_items == self.size:
            first = self.window.popleft()
            self.total -=first
            self.n_of_items -=1
        self.window.append(val)
        self.total += val
        self.n_of_items +=1
        return float(self.total/self.n_of_items)
````

````python
def sortByParity(nums):
  left, right = 0, len(nums)-1
  # two pointer solutions

  while left < right:
    if nums[left] & 1:
      # we are checking odd only and thats why know that right
      # element is correct
      nums[left], nums[right] = nums[right], nums[left]
      right -=1
    else:
      # we are checking even elements that gaurantee only left elements
      left +=1
  return nums

print(sortByParity([3,1,2,4]))
````

## Merge two sorted interval (was asked to Meta phone screen)

````python
# prompt: merge two sorted interval list

def merge_sorted_interval_lists(intervals1, intervals2):
  """Merges two sorted lists of intervals.

  Args:
    intervals1: A list of intervals, where each interval is a list of two
      integers representing the start and end of the interval.
    intervals2: Another list of intervals, similar to intervals1.

  Returns:
    A new list of intervals representing the merged intervals of the two input
    lists.
  """

  merged_intervals = []
  i, j = 0, 0
  while i < len(intervals1) and j < len(intervals2):
    # Find the interval that starts earlier
    if intervals1[i][0] <= intervals2[j][0]:
      current_interval = intervals1[i]
      i += 1
    else:
      current_interval = intervals2[j]
      j += 1

    # Merge with the next interval if they overlap
    while (
        i < len(intervals1)
        and intervals1[i][0] <= current_interval[1]
    ):
      current_interval[1] = max(current_interval[1], intervals1[i][1])
      i += 1
    while (
        j < len(intervals2)
        and intervals2[j][0] <= current_interval[1]
    ):
      current_interval[1] = max(current_interval[1], intervals2[j][1])
      j += 1

    merged_intervals.append(current_interval)

  # Add any remaining intervals from either list
  merged_intervals.extend(intervals1[i:])
  merged_intervals.extend(intervals2[j:])

  return merged_intervals


# Example usage:
intervals1 = [[1, 3], [5, 6], [7, 9]]
intervals2 = [[2, 4], [6, 8]]
merged_intervals = merge_sorted_interval_lists(intervals1, intervals2)
print(merged_intervals)  # Output: [[1, 4], [5, 9]]
````
